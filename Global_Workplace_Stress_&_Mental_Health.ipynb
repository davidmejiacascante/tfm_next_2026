### Estructura general del dataset

El dataset final contiene un total de 162 columnas, que pueden agruparse en dos tipos principales: variables sociodemográficas y variables correspondientes a ítems de encuestas.

Las variables sociodemográficas incluyen información general de los participantes, como edad, género, industria, rol laboral, región, modalidad de trabajo y acceso a servicios de salud mental. Estas variables permiten describir el perfil de las personas encuestadas y comparar resultados entre distintos grupos.

El segundo grupo, que corresponde a la mayoría de las columnas, está formado por preguntas individuales de distintas escalas utilizadas para medir aspectos del entorno laboral y del bienestar de los trabajadores. Estas preguntas recogen información sobre carga de trabajo, demandas emocionales, apoyo social, estrés, bienestar y agotamiento emocional, entre otros factores.

Dado que estas preguntas están pensadas para medir un mismo concepto en conjunto, no se analizarán de forma individual. En su lugar, se agruparán para crear variables resumen por cada escala, lo que facilita el análisis y la interpretación de los resultados.


```python
import re

df_raw = df.copy() #creación de copia del datafrme df

def clean_col(c): #se define una función que se llama clean_col(C). Toma una columna c para hacer una serie de modificaciones.
    c = c.strip() # Elimina cualquier espacio inicial o final del nombre de la columna.
    c = c.replace(" ", "_").replace("-", "_").replace("/", "_") #Reemplaza espacios, guiones y barras diagonales por guiones bajos. Esto ayuda a crear nombres de columna más consistentes y accesibles (p. ej., Rol de trabajo se convierte en rol de trabajo)
    c = re.sub(r"[^0-9a-zA-Z_]+", "", c) #Esta es una sustitución de expresión regular. Busca cualquier carácter que NO sea un número (0-9), una letra mayúscula (A-Z), una letra minúscula (a-z) o un guion bajo (_), y lo reemplaza con una cadena vacía. Esto elimina eficazmente los caracteres especiales.
    c = re.sub(r"_+", "_", c) #Esto reemplaza cualquier secuencia de uno o más guiones bajos con un solo guion bajo, evitando que aparezcan varios guiones bajos juntos.
    return c.lower() #convierte todo el nombre de la columna limpiada a minúsculas.

df = df_raw.copy() #Restablece el df DataFrame a una copia nueva del df_raw original antes de aplicar la limpieza de la columna.
df.columns = [clean_col(c) for c in df.columns] #Esta línea aplica la función clean_col a cada nombre de columna del DataFrame. Itera sobre los nombres de columna existentes, los limpia con la función clean_col y, a continuación, asigna la lista de nombres nuevos y limpios a df.columns.

print(df.columns.tolist()[:40])
print("Total columnas:", len(df.columns))
```

### Tratamiento de valores faltantes (NA)

El análisis inicial de los valores faltantes muestra un total de 810 valores NA distribuidos entre distintas variables. Ninguna columna presenta una cantidad alta de datos faltantes, ya que el máximo observado es de 12 valores por variable.

Este comportamiento es común en encuestas largas, donde algunas personas pueden omitir ciertas preguntas de forma puntual. No se observa que los valores faltantes se concentren en una variable específica ni que sigan un patrón problemático.

Debido a que la cantidad de valores faltantes por columna es baja y las escalas están formadas por varias preguntas, no se eliminan filas ni se aplican métodos de imputación directa. En su lugar, los valores faltantes se manejarán al momento de crear variables resumen por escala, utilizando promedios que ignoran automáticamente los valores NA. De esta forma se conserva la mayor cantidad de información posible sin alterar los datos originales.


```python
na_total = df.isna().sum().sum() #df.isna(): Crea un DataFrame con valores booleanos, donde "True" indica un valor faltante (NaN) y "False" indica un valor válido.sum(): Aplicado a un DataFrame booleano, sum() cuenta los valores verdaderos (es decir, los valores faltantes) de cada columna.sum() (de nuevo): La segunda función sum() suma todos estos recuentos por columna, lo que proporciona el número total de valores faltantes en todo el DataFrame.
print("NA total:", na_total) #Esta línea simplemente imprime el número total calculado de valores faltantes.

na_by_col = df.isna().sum().sort_values(ascending=False) #df.isna().sum(): Como antes, esto calcula la cantidad de valores faltantes para cada columna.sort_values(ascending=False): Esto ordena la serie resultante de valores faltantes en orden descendente, de modo que las columnas con más valores faltantes aparezcan primero.
print("\nTop 20 columnas con más NA:")
print(na_by_col.head(20))
```

### Revisión de filas duplicadas

Se revisó el dataset para identificar posibles filas duplicadas que pudieran afectar los resultados del análisis. El conjunto de datos cuenta con un total de 5000 observaciones y, al comparar las filas completas, no se encontraron registros repetidos.

La eliminación de duplicados no generó cambios en el número de filas, lo que indica que cada observación corresponde a una respuesta única. Por esta razón, no fue necesario aplicar ningún ajuste adicional en este punto.

Este resultado permite continuar con el análisis con la confianza de que los datos no presentan problemas de duplicación.


```python
print("Filas totales:", len(df)) #Esta línea simplemente imprime el número total de filas presentes actualmente en su DataFrame df.
print("Filas únicas:", len(df.drop_duplicates())) #Esta línea crea primero un DataFrame temporal eliminando las filas duplicadas de df (sin modificar df aún). A continuación, calcula e imprime el número de filas en este DataFrame temporal, mostrando así cuántas filas únicas hay.

df = df.drop_duplicates().reset_index(drop=True) #Este es el paso crucial donde se eliminan las filas duplicadas.df.drop_duplicates(): Elimina las filas donde todos los valores de columna son idénticos a los de otra fila. Por defecto, conserva la primera ocurrencia de un conjunto duplicado.reset_index(drop=True): Tras eliminar filas, el índice del DataFrame original podría tener espacios vacíos. Este método restablece el índice a un índice entero predeterminado (0, 1, 2, ...). drop=True garantiza que el índice antiguo no se añada como una nueva columna al DataFrame.

print("Filas luego de drop_duplicates:", len(df)) #Finalmente, esta línea imprime el número total de filas en df después de que se hayan eliminado los duplicados, lo que confirma el nuevo tamaño de su DataFrame.
```

```python
# Este bloque de código está diseñado para identificar y procesar columnas que representan elementos de encuesta individuales (como 'wp1', 'cogn2', etc.) y convertir sus valores en un formato numérico.
item_pattern = re.compile(r"^(auto|wp|cogn|soc|emo|feedb|coach|eri|oc|ss|cs|os|fs|auton|comp|relat|pjf|pof|psf|elt|ee|s|a|d|sf|wb|cors|cas)\d+$")
#Esta línea define un patrón de expresión regular. Busca nombres de columna que comiencen con un prefijo específico (como "auto", "wp", "cogn", etc., que probablemente corresponda a diferentes escalas o categorías psicológicas de su "MAPA DE VARIABLES"), seguido de uno o más dígitos (\d+). Este patrón ayuda a identificar automáticamente todas las columnas que representan preguntas o elementos individuales de la encuesta.
item_cols = [c for c in df.columns if item_pattern.match(c)] #Esta línea utiliza una comprensión de lista para crear una lista llamada item_cols. Itera todos los nombres de columna en su DataFrame df y añade un nombre de columna a item_cols si coincide con la expresión regular item_pattern.
print("Cantidad de ítems detectados:", len(item_cols))
print("Ejemplos:", item_cols[:20])

# Este bucle recorre cada nombre de columna identificado en item_cols. Para cada una de estas columnas:
for c in item_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")
#Esta función intenta convertir los valores de esa columna a un tipo numérico. Si algún valor no se puede convertir (por ejemplo, si es texto o una cadena inesperada), errors="coerce" indica a Pandas que reemplace ese valor no convertible con NaN (No es un número), convirtiendo las entradas no válidas en valores faltantes.
```

### Validación de rangos de respuesta

Durante la revisión de los rangos de las variables numéricas se observó que los ítems de la escala de Cognitive Appraisal (CAS1–CAS12) toman valores entre 0 y 6.

Esto no corresponde a un error en los datos, sino a la forma en que fue definida originalmente esta escala. A diferencia de otras escalas tipo Likert que comienzan en 1, en este caso el valor 0 es una opción válida de respuesta y puede representar la ausencia del aspecto evaluado.

El hecho de que todos los ítems de esta escala compartan el mismo rango de valores indica que los datos son consistentes y están correctamente codificados. Por esta razón, no se aplican transformaciones ni recodificaciones sobre estas variables y se mantiene su escala original para los análisis posteriores.


```python
#Este bloque de código verifica el rango numérico de las columnas de elementos identificadas
if item_cols: #Esto garantiza que las siguientes operaciones solo se ejecuten si se detectan columnas de elementos (es decir, si la lista item_cols no está vacía).
    mins = df[item_cols].min(numeric_only=True)
    maxs = df[item_cols].max(numeric_only=True)
#Estas líneas calculan los valores mínimo y máximo de cada columna de la lista item_cols. numeric_only=True garantiza que solo se consideren las columnas numéricas.
    print("Mínimo global (ítems):", mins.min())
    print("Máximo global (ítems):", maxs.max())
#Estas líneas imprimen los valores mínimo y máximo absolutos encontrados en todas las columnas de elementos. Esto proporciona una idea general del rango total de respuestas en estos elementos de la encuesta.
    # Se inicializa una lista vacía para almacenar información sobre las columnas que podrían contener valores fuera de un rango esperado específico.
    out_of_range = [] #Este bucle itera sobre cada columna del elemento.
    for c in item_cols:
        col_min = df[c].min()
        col_max = df[c].max()
        #Dentro del bucle, encuentra los valores mínimo y máximo específicamente para la columna actual c.
        if pd.notna(col_min) and pd.notna(col_max): #Esta condición comprueba si los valores mínimo y máximo de la columna no son NaN. Esto es importante porque una columna con todos los valores NaN resultaría en NaN para min() y max(), que no se pueden comparar numéricamente.
            if col_min < 1 or col_max > 7: #Si col_min < 1 o col_max > 7: Si el valor mínimo de la columna es menor que 1 o su valor máximo es mayor que 7, significa que la columna contiene valores fuera del rango esperado de 1 a 7 (lo cual suele ser común en las escalas Likert). La condición col_max > 7 es una buena práctica para garantizar que todos los valores estén dentro de los límites esperados.
                out_of_range.append((c, col_min, col_max)) #Si una columna está fuera de rango, su nombre y sus valores mínimo y máximo reales se añaden a la lista "out_of_range".

    print("\nColumnas fuera de rango 1..7 (si aplica):", len(out_of_range))
    print(out_of_range[:20]) #imprimen el recuento de columnas fuera del rango 1-7 y muestran las primeras 20 columnas junto con sus respectivos valores mínimo/máximo.
```

## Construcción de variables agregadas por escala

El dataset contiene una gran cantidad de columnas que corresponden a preguntas individuales de distintos cuestionarios. Muchas de estas preguntas forman parte de una misma escala y están diseñadas para medir un mismo aspecto, como carga laboral, estrés, apoyo social o agotamiento emocional.

Analizar estas preguntas una por una no resulta práctico ni fácil de interpretar. Además, estas preguntas no fueron pensadas para ser analizadas de forma individual, sino como un conjunto que representa una dimensión más amplia.

### Conversión de respuestas tipo Likert a valores numéricos

Antes de poder construir variables agregadas, es necesario convertir las respuestas que actualmente están en formato de texto (por ejemplo, *Never*, *Sometimes*, *Often*, *Very often* o *Strongly agree*) a valores numéricos.

Esta conversión respeta el orden natural de las respuestas. Por ejemplo, respuestas que indican mayor frecuencia o mayor acuerdo reciben valores numéricos más altos, mientras que respuestas que indican menor frecuencia o desacuerdo reciben valores más bajos. Este paso no modifica el significado de los datos, sino que permite trabajar con ellos de forma cuantitativa.

Una vez realizada esta conversión, cada pregunta individual puede representarse mediante un número, lo que permite realizar cálculos como promedios y comparaciones entre personas.

### Construcción de variables resumen por escala

Después de convertir las respuestas a valores numéricos, se procede a construir variables agregadas por escala. Esto se realiza calculando el promedio de las preguntas que pertenecen a una misma dimensión.

Por ejemplo, si una escala de carga laboral está compuesta por varias preguntas, el promedio de esas respuestas representa el nivel general de carga laboral reportado por cada persona. De esta manera, en lugar de trabajar con múltiples columnas separadas, se obtiene una única variable que resume la información principal de la escala.

Este enfoque permite:
- reducir la cantidad de variables del dataset,
- facilitar la interpretación de los resultados,
- y manejar de forma natural los valores faltantes, ya que los promedios se calculan ignorando los valores NA cuando estos son pocos.

### Relación con el análisis de burnout

El objetivo principal de este trabajo es identificar patrones asociados al burnout. Para ello, resulta fundamental contar con variables que representen claramente conceptos como agotamiento emocional, estrés, demandas laborales y apoyo social.

Las variables agregadas por escala permiten comparar grupos de personas, por ejemplo:
- personas con altos niveles de agotamiento emocional frente a aquellas con niveles bajos,
- o analizar cómo se comportan otras variables (carga laboral, apoyo, bienestar) en personas que reportan burnout.

De esta forma, la construcción de variables agregadas es un paso clave para identificar patrones y relaciones que ayuden a comprender qué factores están más asociados al riesgo de burnout en el entorno laboral.


Para evitar modificar el dataset original, todas las transformaciones posteriores se realizarán sobre copias del conjunto de datos. De esta forma, se preserva una versión base de los datos y se garantiza que cada etapa del proceso pueda revisarse o reproducirse de manera independiente.

A partir del dataset original se crearán versiones simplificadas que permitirán convertir las respuestas en texto a valores numéricos y, posteriormente, construir variables agregadas por escala.


```python
import pandas as pd
import numpy as np
df_original = df.copy(deep=True)   # lo guardamos
```

###Crear una copia para convertir texto a números

```python
df_numeric = df_original.copy(deep=True)
```

###Definir mapeos Likert (frecuencia y acuerdo)

```python
# Frecuencia (ej: Never -> Very often)
FREQ_MAP = {
    "never": 1,
    "sometimes": 2,
    "regularly": 3,
    "often": 4,
    "very often": 5,
}

# Acuerdo (ej: Strongly disagree -> Strongly agree)
AGREE_MAP = {
    "strongly disagree": 1,
    "disagree": 2,
    "neutral": 3,
    "agree": 4,
    "strongly agree": 5,
}
```

###Convertir columnas de texto

```python
def try_map_likert_series(s: pd.Series) -> pd.Series:
    """Intenta convertir una columna tipo Likert (texto) a números.
       Si no calza bien, devuelve la columna original."""
    if not pd.api.types.is_object_dtype(s) and not pd.api.types.is_string_dtype(s):
        return s  # no es texto

    # Normalizar texto
    s_norm = s.astype(str).str.strip().str.lower()

    # Si hay strings como "nan", los volvemos NaN reales
    s_norm = s_norm.replace({"nan": np.nan, "none": np.nan, "": np.nan})

    # Valores únicos (sin NaN)
    uniq = set(s_norm.dropna().unique().tolist())
    if len(uniq) == 0:
        return s  # columna vacía

    # Decide si parece frecuencia o acuerdo
    freq_keys = set(FREQ_MAP.keys())
    agree_keys = set(AGREE_MAP.keys())

    # Si TODOS los valores únicos están dentro del mapa -> convertir
    if uniq.issubset(freq_keys):
        return s_norm.map(FREQ_MAP)
    if uniq.issubset(agree_keys):
        return s_norm.map(AGREE_MAP)
        # Si no calza perfecto, no tocamos
    return s
```

###Aplicar la conversión a todas las columnas

```python
converted_cols = []
skipped_cols = []

for c in df_numeric.columns:
    before = df_numeric[c]
    after = try_map_likert_series(before)

    # Detectar si cambió a numérico
    if not after.equals(before) and pd.api.types.is_numeric_dtype(after):
        df_numeric[c] = after
        converted_cols.append(c)
    else:
        skipped_cols.append(c)

print("Columnas convertidas a numérico (texto→número):", len(converted_cols))
print("Ejemplos:", converted_cols[:25])
```
